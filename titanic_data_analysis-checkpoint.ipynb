```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.countplot(x='Survived', data=df)
plt.title("Survival Distribution")
plt.show()
```

More passengers died than survived during the Titanic disaster. The survival rate was significantly lower than the mortality rate.

```python
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title("Gender vs Survival")
plt.show()
```

Female passengers had a much higher survival rate than male passengers. This suggests that women were given priority during evacuation.

```python
sns.countplot(x='Pclass', hue='Survived', data=df)
plt.title("Passenger Class vs Survival")
plt.show()
```

First-class passengers were more likely to survive. Third-class passengers were significantly less likely to survive than the other two upper classes

```python
sns.histplot(df['Age'], bins=30, kde=True)
plt.title("Age Distribution")
plt.show()
```

Most passengers were between 20 and 40 years old.

```python
sns.boxplot(x=df['Fare'])
plt.title("Fare Outliers")
plt.show()
```

The Fare boxplot shows several outliers, indicating that a small number of passengers paid exceptionally high ticket prices compared to the majority of passengers.

```python
numeric_df = df.select_dtypes(include=np.number)
sns.heatmap(numeric_df.corr(), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

Fare and passenger class show noticeable relationships with survival.

# Key Findings

1. Female passengers survived more often than males.
2. First-class passengers had higher survival rates.
3. Higher fares were associated with survival.
4. Most passengers were between 20 and 40 years old.
5. Fare contained several outliers.

# Conclusion

The Titanic dataset analysis revealed that gender, passenger class, and fare were important factors affecting survival. Exploratory Data Analysis helped uncover meaningful patterns and relationships in the dataset.
